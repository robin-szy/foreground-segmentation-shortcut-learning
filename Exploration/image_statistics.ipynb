```python
import os
from pathlib import Path
from PIL import Image
import matplotlib.pyplot as plt
import numpy as np

# set this to your dataset root
DATASET_PATH = Path("/home/manzana/Data_Science/Uni/Deep_Learning_Projects/inaturalist_12K/raw/")

widths, heights, aspect_ratios = [], [], []

# recursively find all images
image_paths = list(DATASET_PATH.rglob("*"))

for path in image_paths:
    if path.suffix.lower() in [".jpg", ".jpeg", ".png"]:
        try:
            with Image.open(path) as img:
                w, h = img.size
                widths.append(w)
                heights.append(h)
                aspect_ratios.append(w / h)
        except:
            continue

print(f"Total images processed: {len(widths)}")

# --- Plot histograms ---
plt.figure()
plt.hist(widths, bins=50)
plt.title("Width Distribution")
plt.xlabel("Width (px)")
plt.ylabel("Count")
plt.show()

plt.figure()
plt.hist(heights, bins=50)
plt.title("Height Distribution")
plt.xlabel("Height (px)")
plt.ylabel("Count")
plt.show()

plt.figure()
plt.hist(aspect_ratios, bins=50)
plt.title("Aspect Ratio Distribution (W/H)")
plt.xlabel("Aspect Ratio")
plt.ylabel("Count")
plt.show()

print("Width  - min/median/max:", np.min(widths), np.median(widths), np.max(widths))
print("Height - min/median/max:", np.min(heights), np.median(heights), np.max(heights))
print("Aspect ratio - min/median/max:", np.min(aspect_ratios), np.median(aspect_ratios), np.max(aspect_ratios))
```
